```python
kata1 = input("Masukkan kata pertama: ")
kata2 = input("Masukkan kata kedua: ")

kata1 = kata1.lower()
kata2 = kata2.lower()

if sorted(kata1) == sorted(kata2):
    print("Anagram")
else:
    print("Bukan Anagram")
```

```python
kalimat = input("Masukkan kalimat: ")
kata = input("Masukkan kata yang dicari: ")

kalimat = kalimat.lower()
kata = kata.lower()

kalimat = kalimat.replace(".", "")

data = kalimat.split()

jumlah = data.count(kata)

print(kata, "ada", jumlah, "buah")
```

```python
kalimat = input("Masukkan kalimat: ")

kalimat = kalimat.lower()

kata = kalimat.split()

terpendek = min(kata, key=len)
terpanjang = max(kata, key=len)

print("terpendek:", terpendek)
print("terpanjang:", terpanjang)
```

```python
import re
from datetime import datetime

kalimat = input("Masukkan kalimat: ")

tanggal_list = re.findall(r'\d{4}-\d{2}-\d{2}', kalimat)

sekarang = datetime.now()

for tgl in tanggal_list:
    tgl_obj = datetime.strptime(tgl, "%Y-%m-%d")

    selisih = (sekarang - tgl_obj).days
    
    print(tgl_obj, "selisih", selisih, "hari")
```

```python
import re
import random
import string

kalimat = input("Masukkan kalimat: ")

emails = re.findall(r'\S+@\S+', kalimat)

for email in emails:
    username = email.split("@")[0]

    password = ''.join(random.choices(string.ascii_letters + string.digits, k=8))
    
    print(email, "username:", username, ", password:", password)
```
